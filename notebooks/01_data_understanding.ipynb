```python
orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"]
)

orders["order_delivered_customer_date"] = pd.to_datetime(
    orders["order_delivered_customer_date"]
)

invalid_delivery_dates = orders[
    orders["order_delivered_customer_date"].notna()
    & (
        orders["order_delivered_customer_date"]
        < orders["order_purchase_timestamp"]
    )
]

print("Delivered before purchase:", len(invalid_delivery_dates))
```

```python
orders["order_approved_at"] = pd.to_datetime(
    orders["order_approved_at"]
)

invalid_approval_dates = orders[
    orders["order_approved_at"].notna()
    & (
        orders["order_approved_at"]
        < orders["order_purchase_timestamp"]
    )
]

print("Approved before purchase:", len(invalid_approval_dates))
```

### Order Approval Date Validation

- No orders were found where payment approval occurred before the order was purchased.
- `order_approved_at` is chronologically valid relative to `order_purchase_timestamp`.
- No records require correction based on this check.

```python
invalid_delivery_sequence = orders[
    orders["order_delivered_customer_date"].notna()
    & orders["order_delivered_carrier_date"].notna()
    & (
        orders["order_delivered_customer_date"]
        < orders["order_delivered_carrier_date"]
    )
]

print("Customer delivery before carrier handover:", len(invalid_delivery_sequence))
```

```python
invalid_delivery_sequence[
    [
        "order_id",
        "order_status",
        "order_purchase_timestamp",
        "order_approved_at",
        "order_delivered_carrier_date",
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ]
]
```

### Delivery Sequence Anomaly

- 23 orders have a customer delivery timestamp earlier than the carrier handover timestamp.
- These records have otherwise plausible order lifecycle dates.
- The anomaly represents approximately 0.02% of all orders.
- The records will not be deleted because the source data should be preserved.
- These records will be flagged as delivery-sequence anomalies and excluded only from analyses where this inconsistent sequence could affect the calculation.

```python
late_deliveries = orders[
    orders["order_delivered_customer_date"].notna()
    & (
        orders["order_delivered_customer_date"]
        > orders["order_estimated_delivery_date"]
    )
]

print("Orders delivered after estimated date:", len(late_deliveries))
```

### Estimated Delivery Date Assessment

- 7,827 delivered orders have a customer delivery date later than the estimated delivery date.
- These records are not considered data-quality errors because late delivery is a valid business outcome.
- Late delivery will be retained as a business metric for operational analysis.
- This metric can be used to evaluate delivery performance and its relationship with customer satisfaction.
